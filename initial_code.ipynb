```python
#read in schedule from Word document
#load Word document and create Document object called doc
doc = Document("schedule_grid.docx")

#access tables in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create empty list to capture data
data1 = []
#for loop to get rows in table1 to put data into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create empty list to capture data
data2 = []
#for loop to get rows in table2 to put data into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#organize data from data list into dataframe
#create list for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Game_Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with information from data list with nested for loop
#for each element (game) in data
for element in data:
    #for each index and value in the element (game details)
    for index,x in enumerate(element):
        #if index=0 append value to Data list
        if index == 0:
            Date.append(x)
        #if index=1 append value to Time list
        elif index == 1:
            Time.append(x)
        #if index=2 append value to At list
        elif index == 2:
            At.append(x)
        #if index=3 append value to Opponent list
        elif index == 3:
            Opponent.append(x)
        #if index=4 append value to Location list
        elif index == 4:
            Game_Location.append(x)
        #if index=5 append value to TV list
        elif index == 5:
            TV.append(x)
        #if index=6 append value to Radio list
        elif index == 6:
            Radio.append(x)
        #if index=7 append value to Tournament list
        elif index == 7:
            Tournament.append(x)
        #if index=8 append value to Result list
        else:
            Result.append(x)
```

```python
# geocoding Game_Location column for scattermap
#use jittering to create unique coordinates so that 2 games with the same location do not have the exact same coordinates
Game_Longitude=[]
Game_Latitude=[]
jitter_amount = 0.0001

geolocator = Nominatim(user_agent="my_app")
#wrapping geolocator function in Ratelimiter function with 1 second delay to avoid error messages about too many requests
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)
for x in df['Game_Location']:
	location = geocode(x)
	Game_Latitude.append(location.latitude + np.random.uniform(-jitter_amount, jitter_amount))
	Game_Longitude.append(location.longitude + np.random.uniform(-jitter_amount, jitter_amount))
```

```python
# geocoding Opponent column 
#use jittering to create unique coordinates so that 2 games with the same location do not have the exact same coordinates
Opponent_Latitude = []
Opponent_Longitude = []

jitter_amount = 0.0001

geolocator = Nominatim(user_agent="opp_locator_app")
#wrapping geolocator function in Ratelimiter function with 1 second delay to avoid error messages about too many requests
geocode1 = RateLimiter(geolocator.geocode, min_delay_seconds=1)
for x in df['Opponent']:
	opp_locate = geocode1(x)
	Opponent_Latitude.append(opp_locate.latitude + np.random.uniform(-jitter_amount, jitter_amount))
	Opponent_Longitude.append(opp_locate.longitude + np.random.uniform(-jitter_amount, jitter_amount))
```
