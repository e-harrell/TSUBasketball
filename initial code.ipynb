### Schedule Data

```python
#read in schedule from Word document
#load Word document
doc = Document("schedule_grid.docx")

#access tables in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create list to capture data
data1 = []
#for loop to get data from table1 rows into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create list to capture data
data2 = []
#for loop to get data from table2 rows into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#checking the length of the lists
len(data1)
len(data2)
```

```python
#checking beginning and end of data2 to make sure it captured everything
data2[0]
data2[-1]
```

```python
#check type and length of each element in data1 and data2
for x in data1:
    print("data type:", type(x), "and length:", len(x))
for x in data2:
    print("data type:", type(x), "and length:", len(x))
```

```python
#organize data from data1 and data2
#create lists for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with data from data1
for element in data1:
    for index,x in enumerate(element):
        if index == 0:
            Date.append(x)
        elif index ==1:
            Time.append(x)
        elif index == 2:
            At.append(x)
        elif index == 3:
            Opponent.append(x)
        elif index == 4:
            Location.append(x)
        elif index == 5:
            TV.append(x)
        elif index == 6:
            Radio.append(x)
        elif index == 7:
            Tournament.append(x)
        else:
            Result.append(x)
```

```python
#attaching data from data2 to above lists
for element in data2:
    for index,x in enumerate(element):
        if index == 0:
            Date.append(x)
        elif index ==1:
            Time.append(x)
        elif index == 2:
            At.append(x)
        elif index == 3:
            Opponent.append(x)
        elif index == 4:
            Location.append(x)
        elif index == 5:
            TV.append(x)
        elif index == 6:
            Radio.append(x)
        elif index == 7:
            Tournament.append(x)
        else:
            Result.append(x)
```

```python
#putting lists into dataframe
df = pd.DataFrame()
df['Date']=Date
df['Time']=Time
df['At']=At
df['Opponent']=Opponent
df['Location']=Location
df['TV']=TV
df['Radio']=Radio
df['Tournament']=Tournament
df['Result']=Result
```

```python
#looking at dataframe and getting information about dataframe
df.head()
df.tail()
df.info()
```

```python
#splitting Date column 
df[["Date", "Dayofweek"]] = df.Date.str.split('(', expand=True)
#remove ) from Dayofweek column
df['Dayofweek']=df['Dayofweek'].str.replace(')', '')
#rearrange columns
col = df.pop('Dayofweek')
df.insert(1, 'Dayofweek', col)
df.head()
```
