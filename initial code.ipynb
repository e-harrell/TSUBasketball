### Schedule Data

```python
#read in schedule from Word document
#load Word document and create Document object called doc
doc = Document("schedule_grid.docx")

#access tables in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create empty list to capture data
data1 = []
#for loop to get rows in table1 to put data into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create empty list to capture data
data2 = []
#for loop to get rows in table2 to put data into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#combine data1 and data2
data = data1 + data2
print(data)
len(data)
type(data)
```

```python
#organize data from data into dataframe
#create list for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with data from data with nested for loop
#for each list (game) in data
for element in data:
    #for each index and element in list (game)
    for index,x in enumerate(element):
        #if index=0 append element to Data list
        if index == 0:
            Date.append(x)
        #if index=1 append element to Time list
        elif index == 1:
            Time.append(x)
        #if index=2 append element to At list
        elif index == 2:
            At.append(x)
        #if index=3 append element to Opponent list
        elif index == 3:
            Opponent.append(x)
        #if index=4 append element to Location list
        elif index == 4:
            Location.append(x)
        #if index=5 append element to TV list
        elif index == 5:
            TV.append(x)
        #if index=6 append element to Radio list
        elif index == 6:
            Radio.append(x)
        #if index=7 append element to Tournament list
        elif index == 7:
            Tournament.append(x)
        #if index=8 append element to Result list
        else:
            Result.append(x)
```

```python
#drop Radio and Tournament columns
df.columns
df=df.drop(columns=['Radio', 'Tournament'])
df.columns
```

```python
#look at dataframe
df.head()
df.tail()
df.info()
```

```python
#splitting Date column into Date and Dayofweek columns
df[["Date", "Dayofweek"]] = df.Date.str.split('(', expand=True)
#remove ) from Dayofweek column
df['Dayofweek']=df['Dayofweek'].str.replace(')', '')
#rearrange columns
col = df.pop('Dayofweek')
df.insert(1, 'Dayofweek', col)
df.head()
```

```python
#splitting out Location column into City and State columns
df[["City", "State"]] = df.Location.str.split(',', expand=True)
#check df
df.head()
#drop Location
df=df.drop(columns=["Location"])
df.head()
```

```python
#split out Result column into WinLoss, TSUscore, Oppscore
#first split Result into WinLoss & Result
df[["WinLoss", "Result"]] = df.Result.str.split(',', expand=True)
#make WinLoss be all uppercase
df['WinLoss']= df['WinLoss'].str.upper()
#split Result again into TSUscore and Oppscore
df[['TSUscore', 'Oppscore']] = df.Result.str.split('-', expand=True)
#drop Result column
df=df.drop(columns=['Result'])
df.head()
```

```python
#get information on dataframe
df.info()
```

```python
#correcting datatypes of columns
df['Date']=pd.to_datetime(df['Date'])
df['TSUscore']=pd.to_numeric(df['TSUscore'])
df['Oppscore']=pd.to_numeric(df['Oppscore'])
df.head()
df.info()
```

```python
#rename At column to HomeAway
df = df.rename(columns={"At": "HomeAway"})
df.head()
df.info()
```
