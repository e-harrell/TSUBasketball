### Schedule Data

```python
#read in schedule from Word document
#load Word document and create Document object called doc
doc = Document("schedule_grid.docx")

#access tables in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create empty list to capture data
data1 = []
#for loop to get rows in table1 to put data into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create empty list to capture data
data2 = []
#for loop to get rows in table2 to put data into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#organize data from data1 and data2 into dataframe
#create list for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with data from data1 with nested for loop
#for each list (game) in data1
for element in data1:
    #for each index and element in list (game)
    for index,x in enumerate(element):
        #if index=0 append element to Data list
        if index == 0:
            Date.append(x)
        #if index=1 append element to Time list
        elif index == 1:
            Time.append(x)
        #if index=2 append element to At list
        elif index == 2:
            At.append(x)
        #if index=3 append element to Opponent list
        elif index == 3:
            Opponent.append(x)
        #if index=4 append element to Location list
        elif index == 4:
            Location.append(x)
        #if index=5 append element to TV list
        elif index == 5:
            TV.append(x)
        #if index=6 append element to Radio list
        elif index == 6:
            Radio.append(x)
        #if index=7 append element to Tournament list
        elif index == 7:
            Tournament.append(x)
        #if index=8 append element to Result list
        else:
            Result.append(x)
```

```python
#attaching data from data2 to above lists
for element in data2:
    for index,x in enumerate(element):
        if index == 0:
            Date.append(x)
        elif index ==1:
            Time.append(x)
        elif index == 2:
            At.append(x)
        elif index == 3:
            Opponent.append(x)
        elif index == 4:
            Location.append(x)
        elif index == 5:
            TV.append(x)
        elif index == 6:
            Radio.append(x)
        elif index == 7:
            Tournament.append(x)
        else:
            Result.append(x)
```

```python
#check length of lists
len(Date)
len(Time)
len(At)
len(Opponent)
len(Location)
len(TV)
len(Radio) 
len(Tournament)
len(Result)
```

```python
#checking that data went to the correct lists
print(Date)
print(Time)
print(At)
print(Opponent)
print(Location)
print(TV)
print(Radio) 
print(Tournament)
print(Result)
```

```python
#clean Time list
print(Time)
Time = [s.replace(' CT', '') for s in Time]
Time[9] ='8:30 PM'
print(Time)
```

```python
#clean Opponent list
print(Opponent)
Opponent = [a.replace('AlabamaA&M', 'Alabama A&M') for a in Opponent]
Opponent = [b.replace('SEMO', 'Southeast Missouri') for b in Opponent]
Opponent = [c.replace('SIUE', 'Southeastern Illinois University Edawrdsville') for c in Opponent]
Opponent = [d.replace('Alabama A&M', '') for d in Opponent]
print(Opponent)
```

```python
#clean Location
print(Location)
for index,element in enumerate(Location):
    if element.contains('NASHVILLE'):
        Location[index] == 'Nashville, Tenn.'
    else:
        continue
print(Location)
```

```python

#clean Result
print(Result)
Result = [s.replace('\n', '') for s in Result]
print(Result)
```

```python
#putting lists into dataframe
df = pd.DataFrame()
df['Date']=Date
df['Time']=Time
df['At']=At
df['Opponent']=Opponent
df['Location']=Location
df['TV']=TV
df['Radio']=Radio
df['Tournament']=Tournament
df['Result']=Result
```

```python
#looking at dataframe and getting information about dataframe
df.head()
df.tail()
df.info()
```

```python
#splitting Date column 
df[["Date", "Dayofweek"]] = df.Date.str.split('(', expand=True)
#remove ) from Dayofweek column
df['Dayofweek']=df['Dayofweek'].str.replace(')', '')
#rearrange columns
col = df.pop('Dayofweek')
df.insert(1, 'Dayofweek', col)
df.head()
```
