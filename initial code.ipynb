### Schedule Data

```python
#read in schedule from Word document
#load Word document
doc = Document("schedule_grid.docx")

#access 1st table in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create list to capture data
data1 = []
#for loop to get data into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create list to capture data
data2 = []
#for loop to get data into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#create lists for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with data from data1
for element in data1:
    for index,x in enumerate(element):
        if index == 0:
            Date.append(x)
        elif index ==1:
            Time.append(x)
        elif index == 2:
            At.append(x)
        elif index == 3:
            Opponent.append(x)
        elif index == 4:
            Location.append(x)
        elif index == 5:
            TV.append(x)
        elif index == 6:
            Radio.append(x)
        elif index == 7:
            Tournament.append(x)
        else:
            Result.append(x)
```

```python
#attaching data from data2
for element in data2:
    for index,x in enumerate(element):
        if index == 0:
            Date.append(x)
        elif index ==1:
            Time.append(x)
        elif index == 2:
            At.append(x)
        elif index == 3:
            Opponent.append(x)
        elif index == 4:
            Location.append(x)
        elif index == 5:
            TV.append(x)
        elif index == 6:
            Radio.append(x)
        elif index == 7:
            Tournament.append(x)
        else:
            Result.append(x)
```

```python
#putting lists into dataframe
df = pd.DataFrame()
df['Date']=Date
df['Time']=Time
df['At']=At
df['Opponent']=Opponent
df['Location']=Location
df['TV']=TV
df['Radio']=Radio
df['Tournament']=Tournament
df['Result']=Result
```
