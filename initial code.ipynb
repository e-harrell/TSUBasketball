### Schedule Data

```python
#read in schedule from Word document
#load Word document and create Document object called doc
doc = Document("schedule_grid.docx")

#access tables in doc
table1 = doc.tables[1]
table2 = doc.tables[2]

#extract data from 1st table:
#create empty list to capture data
data1 = []
#for loop to get rows in table1 to put data into list
for row in table1.rows:
    data1.append([cell.text for cell in row.cells])


#extract data from 2nd table:
#create empty list to capture data
data2 = []
#for loop to get rows in table2 to put data into list
for row in table2.rows:
    data2.append([cell.text for cell in row.cells])
```

```python
#combine data1 and data2
data = data1 + data2
print(data)
len(data)
type(data)
```

```python
#organize data from data list into dataframe
#create list for each column in data frame
Date=[]
Time =[]
At =[]
Opponent = []
Location = []
TV = []
Radio = []
Tournament = []
Result=[]
#fill lists with information from data list with nested for loop
#for each element (game) in data
for element in data:
    #for each index and value in the element (game details)
    for index,x in enumerate(element):
        #if index=0 append value to Data list
        if index == 0:
            Date.append(x)
        #if index=1 append value to Time list
        elif index == 1:
            Time.append(x)
        #if index=2 append value to At list
        elif index == 2:
            At.append(x)
        #if index=3 append value to Opponent list
        elif index == 3:
            Opponent.append(x)
        #if index=4 append value to Location list
        elif index == 4:
            Location.append(x)
        #if index=5 append value to TV list
        elif index == 5:
            TV.append(x)
        #if index=6 append value to Radio list
        elif index == 6:
            Radio.append(x)
        #if index=7 append value to Tournament list
        elif index == 7:
            Tournament.append(x)
        #if index=8 append value to Result list
        else:
            Result.append(x)
```

```python
#splitting Date column into Date and Dayofweek columns
df[["Date", "Dayofweek"]] = df.Date.str.split('(', expand=True)
#remove ) from Dayofweek column
df['Dayofweek']=df['Dayofweek'].str.replace(')', '')
#rearrange columns
#create col series by  removing Dayofweek column
col = df.pop('Dayofweek')
#insert Dayofweek column as column in index 1 of dataframe
df.insert(1, 'Dayofweek', col)
#look at rearranged dataframe
df.head()
```

```python
#split Result column into WinLoss, TSUscore, Oppscore columns
#first split Result into WinLoss & Result
df[["WinLoss", "Result"]] = df.Result.str.split(',', expand=True)
#make WinLoss uppercase
df['WinLoss']= df['WinLoss'].str.upper()
#split Result again into TSUscore and Oppscore
df[['TSUscore', 'Oppscore']] = df.Result.str.split('-', expand=True)
#drop Result column
df=df.drop(columns=['Result'])
df.head()
```

```python
#correcting datatypes of columns
df['Date']=pd.to_datetime(df['Date'])
df['TSUscore']=pd.to_numeric(df['TSUscore'])
df['Oppscore']=pd.to_numeric(df['Oppscore'])
#look at dataframe and info
df.head()
df.tail()
df.info()
```
